```python
import pybryt
```

```python
def value(x,y):
    pybryt.Value(x,success_message="讀取蘋果價格",failure_message="請輸入正確數字")
    pybryt.Value(y,success_message="讀取數量",failure_message="請輸入正確數字")
    if (x == 0 or y==0):
        return "錯誤"
    else:
        return x*y
```

```python
val=value(26,6)
val
```

```python
pybryt.Value(value(26,6),success_message="價格計算正確",failure_message="請輸入正確訊息")
```

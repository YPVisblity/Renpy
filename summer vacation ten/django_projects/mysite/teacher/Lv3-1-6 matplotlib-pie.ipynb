```python
import pybryt
```

```python
count = 4
labels = ["睡覺", "上課", "寫作業", "休閒"]
title = "一天時間分配"
```

```python
count_results = ("chapter-3-level-1-4:count", count)
labels_results = ("chapter-3-level-1-4:labels", labels)
title_results = ("chapter-3-level-1-4:title", title)

pybryt.Value(count_results, name="區塊數量",
             success_message="圓餅圖區塊數量正確",
             failure_message="請使用 plt.pie() 畫出 4 個區塊")

pybryt.Value(labels_results, name="分類標籤",
             success_message="分類標籤正確",
             failure_message="圓餅圖標籤應該是 睡覺、上課、寫作業、休閒")

pybryt.Value(title_results, name="標題",
             success_message="標題通過",
             failure_message="標題應該是 一天時間分配")
```

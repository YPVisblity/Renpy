```python
import pybryt
```

```python
def grade(score):
    if score >= 90:
        return "A"
    elif score >= 80:
        return "B"
    elif score >= 70:
        return "C"
    elif score >= 60:
        return "D"
    else:
        return "F"

pybryt.ReturnValue(grade(95), success_message="grade 95 passed", failure_message="grade 95 should return A")
pybryt.ReturnValue(grade(90), success_message="grade 90 passed", failure_message="grade 90 should return A")
pybryt.ReturnValue(grade(86), success_message="grade 86 passed", failure_message="grade 86 should return B")
pybryt.ReturnValue(grade(80), success_message="grade 80 passed", failure_message="grade 80 should return B")
pybryt.ReturnValue(grade(72), success_message="grade 72 passed", failure_message="grade 72 should return C")
pybryt.ReturnValue(grade(70), success_message="grade 70 passed", failure_message="grade 70 should return C")
pybryt.ReturnValue(grade(64), success_message="grade 64 passed", failure_message="grade 64 should return D")
pybryt.ReturnValue(grade(60), success_message="grade 60 passed", failure_message="grade 60 should return D")
pybryt.ReturnValue(grade(40), success_message="grade 40 passed", failure_message="grade 40 should return F")
```

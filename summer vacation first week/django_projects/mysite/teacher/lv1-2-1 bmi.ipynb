```python
import pybryt
```

```python
import math
def basic(height,weight):

    bmi= weight/pow(height,2)
    if (bmi < 18.5):
        return "過輕"
    elif (bmi >24):
        return "過重"
    else:
        return "正常"
    
pybryt.ReturnValue(basic(1.78, 70),success_message="正常",failure_message="缺少過輕或過重條件") #正常
pybryt.ReturnValue(basic(1.60, 40),success_message="過輕",failure_message="缺少正常或過重條件") #過輕
pybryt.ReturnValue(basic(1.60, 80),success_message="過重",failure_message="缺少正常或過輕條件") #過重
```

```python
result = basic(1.78, 70)
result
```

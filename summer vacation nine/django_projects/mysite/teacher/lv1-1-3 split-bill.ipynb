```python
import pybryt
```

```python
def split_bill(price, quantity, people):
    total = price*quantity
    return round(total/people,2)
```

```python
result1 = split_bill(45.6,3,3)   # 45.6*3/3 = 45.6
result2 = split_bill(100,7,4)    # 100*7/4 = 175.0
result3 = split_bill(58.9,2,5)   # 58.9*2/5 = 23.56

pybryt.Value(result1, name="平分帳單(45.6,3,3)",
             success_message="計算與四捨五入正確",
             failure_message="應回傳 45.6")

pybryt.Value(result2, name="平分帳單(100,7,4)",
             success_message="不同數值測試正確",
             failure_message="應回傳 175.0")

pybryt.Value(result3, name="平分帳單(58.9,2,5)",
             success_message="小數精度處理正確",
             failure_message="應回傳 23.56（注意四捨五入到小數點後2位）")
```

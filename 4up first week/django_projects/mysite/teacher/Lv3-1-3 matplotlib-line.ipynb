```python
import pybryt
```

```python
line_count = 1
x_data = [1, 2, 3, 4, 5]
y_data = [22, 24, 23, 26, 28]
plot_title = "溫差"
x_label = "天數"
y_label = "溫度"
```

```python
line_count_signature = ("chapter-3-level-1-1:line_count", line_count)
x_data_signature = ("chapter-3-level-1-1:x_data", x_data)
y_data_signature = ("chapter-3-level-1-1:y_data", y_data)
title_signature = ("chapter-3-level-1-1:title", plot_title)
xlabel_signature = ("chapter-3-level-1-1:xlabel", x_label)
ylabel_signature = ("chapter-3-level-1-1:ylabel", y_label)

pybryt.Value(line_count_signature, name="折線數量",
             success_message="已畫出一條折線圖",
             failure_message="請使用 plt.plot() 畫出一條折線圖")

pybryt.Value(x_data_signature, name="x軸資料",
             success_message="x 軸資料正確",
             failure_message="x 軸資料應該是 [1, 2, 3, 4, 5]")

pybryt.Value(y_data_signature, name="y軸資料",
             success_message="y 軸資料正確",
             failure_message="y 軸資料應該是 [22, 24, 23, 26, 28]")

pybryt.Value(title_signature, name="圖表標題",
             success_message="圖表標題正確",
             failure_message="標題應該是 溫差")

pybryt.Value(xlabel_signature, name="x軸名稱",
             success_message="x 軸名稱正確",
             failure_message="x 軸名稱應該是 天數")

pybryt.Value(ylabel_signature, name="y軸名稱",
             success_message="y 軸名稱正確",
             failure_message="y 軸名稱應該是 溫度")
```

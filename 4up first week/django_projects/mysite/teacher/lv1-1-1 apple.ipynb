```python
import pybryt
```

```python

def value(x,y):
    if (x == 0 or y==0):
        return "ERROR"
    else:
        return x*y
```

```python
result1 = value(26, 6)      #多組測資
result2 = value(10, 3)      #多組測資
result3 = value(7, 8)       #多組測資
result4 = value(5, 0)       #測試為0的情況
```

```python
pybryt.Value(result1,success_message="蘋果價格計算正確",failure_message="value(26,6) 應回傳 156")
pybryt.Value(result2,success_message="10 * 3 計算正確",failure_message="value(10,3) 應回傳 30")
pybryt.Value(result3,success_message="7 * 8 計算正確",failure_message="value(7,8) 應回傳 56")
pybryt.Value(result4,success_message="正確處理為0情況",failure_message="value(5,0) 應回傳 錯誤 而非 0")
```

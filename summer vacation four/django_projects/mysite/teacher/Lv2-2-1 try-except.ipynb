```python
import pybryt
import numpy as np
```

```python
def safe_divide(a, b):
    try:
        return a/b
    except:
        return "Error:除以零"

val1=safe_divide(10,0)
val2=safe_divide(10,2)


```

```python
pybryt.Value(val1, name="除以零錯誤處理",success_message="except方法正確",failure_message="當 b=0 時應回傳 'Error:除以零'")
pybryt.Value(val2, name="正常除法結果",success_message="a除以b正確",failure_message="safe_divide(10, 2) 應回傳 5.0")
```

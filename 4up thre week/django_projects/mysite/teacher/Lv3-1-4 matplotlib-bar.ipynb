```python
import pybryt
```

```python
count = 4
price = [30, 45, 25, 40]
fruit = ["Apple", "Banana", "Orange", "Grape"]
```

```python
title = "水果價格"
xlabel = "水果"
ylabel = "價錢"

count_results = ("chapter-3-level-1-2:count", count)
price_results = ("chapter-3-level-1-2:price", price)
fruit_results = ("chapter-3-level-1-2:fruit", fruit)
title_results = ("chapter-3-level-1-2:title", title)
xlabel_results = ("chapter-3-level-1-2:xlabel", xlabel)
ylabel_results = ("chapter-3-level-1-2:ylabel", ylabel)

pybryt.Value(count_results, name="長條數量",
             success_message="長條圖測試",
             failure_message="請使用 plt.bar() 畫出長條圖")

pybryt.Value(fruit_results, name="x軸分類",
             success_message="x 軸通過",
             failure_message="x 軸上應該要是 [Apple, Banana, Orange, Grape]")

pybryt.Value(price_results, name="長條高度",
             success_message="y 軸通過",
             failure_message="y 軸上應該要是 [30, 45, 25, 40]")

pybryt.Value(title_results, name="標題",
             success_message="標題通過",
             failure_message="標題錯誤")

pybryt.Value(xlabel_results, name="x軸名稱",
             success_message="x 軸名稱正確",
             failure_message="x 軸名稱錯誤")

pybryt.Value(ylabel_results, name="y軸名稱",
             success_message="y 軸名稱正確",
             failure_message="y 軸名稱錯誤")
```

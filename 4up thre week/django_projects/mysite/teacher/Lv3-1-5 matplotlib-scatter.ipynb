```python
import pybryt
```

```python
count = 5
points = [(150.0, 45.0), (160.0, 55.0), (165.0, 60.0), (170.0, 68.0), (180.0, 75.0)]
title = "身高體重關係"
xlabel = "身高"
ylabel = "體重"
```

```python
count_results = ("chapter-3-level-1-3:count", count)
points_results = ("chapter-3-level-1-3:points", points)
title_results = ("chapter-3-level-1-3:title", title)
xlabel_results = ("chapter-3-level-1-3:xlabel", xlabel)
ylabel_results = ("chapter-3-level-1-3:ylabel", ylabel)

pybryt.Value(count_results, name="點的數量",
             success_message="散點圖測試通過",
             failure_message="請使用 plt.scatter() 畫出 5 個點")

pybryt.Value(points_results, name="座標資料",
             success_message="座標資料正確",
             failure_message="散點圖座標應該符合 height 和 weight 的資料")

pybryt.Value(title_results, name="標題",
             success_message="標題通過",
             failure_message="標題應該是 身高體重關係")

pybryt.Value(xlabel_results, name="x軸名稱",
             success_message="x 軸名稱正確",
             failure_message="x 軸名稱應該是 身高")

pybryt.Value(ylabel_results, name="y軸名稱",
             success_message="y 軸名稱正確",
             failure_message="y 軸名稱應該是 體重")
```
